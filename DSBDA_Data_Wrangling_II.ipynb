# DSBDA Practical – Data Wrangling II

Academic Performance Dataset preprocessing using Python.

## Step 1: Import Required Libraries

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Step 2: Create Academic Performance Dataset

```python

data = {
    'Student_ID': [1, 2, 3, 4, 5, 6, 7],
    'Math_Score': [78, 85, np.nan, 95, 120, 67, 80],
    'Attendance': [90, 85, 88, np.nan, 92, 40, 89],
    'Study_Hours': [2, 3, 4, 10, 12, 1, 3]
}

df = pd.DataFrame(data)
df
```

## Step 3: Missing Values and Inconsistencies

```python

df.isnull().sum()
```

```python
df['Math_Score'] = df['Math_Score'].fillna(df['Math_Score'].mean())
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].mean())
df
```

## Step 4: Outlier Detection (IQR Method)

```python

Q1 = df['Math_Score'].quantile(0.25)
Q3 = df['Math_Score'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

lower, upper
```

```python

df['Math_Score'] = np.where(df['Math_Score'] > upper, upper, df['Math_Score'])
df
```

## Step 5: Data Transformation (Log Transformation)

```python

df['Study_Hours_Log'] = np.log(df['Study_Hours'] + 1)
df
```

## Conclusion


- Missing values handled using mean  
- Outliers treated using IQR method  
- Log transformation applied to reduce skewness
